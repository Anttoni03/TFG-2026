## 4. Inferencia sobre un Vídeo
Para los vídeos, el modelo procesará frame a frame.

```python
video_prueba = 'video_test.mp4'

print(f"Iniciando el procesamiento del vídeo '{video_prueba}'...")

# Ejecutamos la predicción (puede tardar dependiendo de la CPU/GPU)
resultados_vid = model.predict(source=video_prueba, save=True, conf=0.35, stream=True)

carpeta_salida_vid = ""

# Iteramos sobre el generador. Esto obliga al modelo a procesar frame a frame.
for frame_resultado in resultados_vid:
    carpeta_salida_vid = frame_resultado.save_dir

print(f"Vídeo procesado con éxito.")
print(f"📁 Ve a la carpeta de tu ordenador '{carpeta_salida_vid}' para reproducir el vídeo con las detecciones.")
```
